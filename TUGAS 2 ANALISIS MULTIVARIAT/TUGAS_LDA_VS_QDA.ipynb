```python
import pandas as pd
import numpy as np

# Import dataset dari lokal
DATA_PATH = r"D:\College\Matana\Semester 5\Analisis Multivariat\star_classification.csv"
cols = ["alpha", "delta", "u", "g", "r", "i", "z"]
data = pd.read_csv(DATA_PATH, usecols=cols + ["class"])
# Pisahkan variabel prediktor (X) dan variabel respon (y)
X = data.drop(columns=["class"])
y = data["class"]

# Cek ukuran data
print(X.shape)

# Cek jumlah masing-masing kelas
print(y.value_counts().sort_index())

print(X.columns.tolist())
print(X.describe().T[["mean","min","max"]])
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.30,
    random_state=42,
    stratify=y
)
print(X_train.shape, X_test.shape)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.discriminant_analysis import (
    LinearDiscriminantAnalysis,
    QuadraticDiscriminantAnalysis
)

lda_model = Pipeline([
    ('scaler', StandardScaler()),
    ('model', LinearDiscriminantAnalysis())
])
qda_model = Pipeline([
    ('scaler', StandardScaler()),
     ('model', QuadraticDiscriminantAnalysis())
])
```

```python
lda_model.fit(X_train, y_train)
qda_model.fit(X_train, y_train)

lda_pred = lda_model.predict(X_test)
qda_pred = qda_model.predict(X_test)
```

```python
from sklearn.metrics import confusion_matrix

lda_cm = confusion_matrix(y_test, lda_pred)
qda_cm = confusion_matrix(y_test, qda_pred)

print('LDA')
print(lda_cm)
print('QDA')
print(qda_cm)
```

```python
from sklearn.metrics import classification_report
print('LDA REPORT')
print(classification_report(y_test, lda_pred))
print('QDA REPORT')
print(classification_report(y_test, qda_pred))
```

```python
from sklearn.model_selection import (
    StratifiedKFold, cross_val_score
)
cv = StratifiedKFold(
    n_splits=5, shuffle=True,
    random_state=42)

lda_scores = cross_val_score(
    lda_model, X, y, cv=cv,
    scoring='accuracy')
qda_scores = cross_val_score(
    qda_model, X, y, cv=cv,
    scoring='accuracy')

```

```python
print(lda_scores)
print(qda_scores)
```
